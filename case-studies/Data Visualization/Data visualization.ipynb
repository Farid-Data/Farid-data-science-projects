```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
from sklearn.preprocessing import OneHotEncoder, LabelEncoder
```

```python
tips = sns.load_dataset("tips")
tips.head(5)
```

```python
le = LabelEncoder()
tips["Target"] = le.fit_transform(tips["smoker"])
```

```python
le = LabelEncoder()
tips["sex_target"] = le.fit_transform(tips["sex"])
tips[["sex", "sex_target"]] 
```

```python
sex_map = {"Male":0 , "Female":1}
tips["sex_target_map"] = tips["sex"].map(sex_map)
```

```python
smoker_map = {"No": 1 , "Yes":0}
tips["smoker_map_target"] = tips["smoker"].map(smoker_map)
```

```python
# tips.drop(columns=["sex_reversed_target"], inplace=True)
tips.head(5)
```

```python
smoker_map = {
    "No": 0,
    "Yes": 1
}
tips["smoke_Target"] = tips["smoker"].map(smoker_map)
```

```python
tips["time"].value_counts()


time_map = {"Lunch":0, "Dinner":1}


tips["time_target"] = tips["time"].map(time_map)
```

```python
tips["time_target_le"] = le.fit_transform(tips["time"])
```

```python
tips["smoker_map_target_le"] = le.fit_transform(tips["time"])
```

```python
# Let's DO Some Revision:

fig, axes = plt.subplots(
    3, #    3 rows
    3, #    3 Columns
    figsize=(20, 20),           # Length and width
    constrained_layout = True    # space around charts controlled automatically
)

axes = axes.flatten()


axes[0].hist(
    
    tips["total_bill"],
    bins = 30,
    color = "skyblue",
    edgecolor = "black",
    density = True)
axes[0].set_title("Histogram of Total Bill")
axes[0].set_xlabel("Total Bill ($)")
axes[0].set_ylabel("Density")
axes[0].grid(True)


smoke_count = tips.groupby("smoker")["total_bill"].mean()

axes[1].bar(smoke_count.index,
            smoke_count.values,
            color="red"
    
)
axes[1].set_title("Average Total Bill by Smoker")
axes[1].set_xlabel("Smoker")
axes[1].set_ylabel("Average Bill ($)")



day_count_me = tips.groupby("day")["total_bill"].mean()

axes[2].pie(
    day_count_me,
    labels = day_count_me.index,
    startangle = 90,
    autopct = "%1.1f%%",
    explode = [0, 0, 0.2, 0]
)
axes[2].set_title("Average Tips by Day")




male = tips[tips["sex"] == "Male"]
female = tips[tips["sex"] == "Female"]

axes[3].scatter(
    male.index,
    male["total_bill"],
    c="blue",
    s=male["size"] * 10,
    alpha=0.7,
    label="Male"
)

axes[3].scatter(
    female.index,
    female["total_bill"],
    c="red",
    s=female["size"] * 10,
    alpha=0.7,
    label="Female"
)

axes[3].set_title("Total Bill by Gender")
axes[3].set_xlabel("Record Index")
axes[3].set_ylabel("Total Bill ($)")

axes[3].legend()



sns.heatmap(tips.corr(numeric_only = True),
            cmap="Blues",
            annot = True,
            ax = axes[4])

axes[4].set_title("Correlation Heatmap")



axes[5].plot(
    tips.index,
    tips["total_bill"],
    color="black",
    linewidth=0.8,
    label="Total Bill"
)

axes[5].fill_between(
    tips.index,
    tips["total_bill"],
    color="skyblue",
    alpha=0.7
)

axes[5].set_title("Total Bill Area Plot")
axes[5].set_xlabel("Record Index")
axes[5].set_ylabel("Total Bill ($)")

axes[5].legend()




sns.violinplot(data=tips,
               x = "day",
               y="total_bill",
               ax=axes[6])

plt.title("Total Bill Distribution by Day")







plt.show()
```

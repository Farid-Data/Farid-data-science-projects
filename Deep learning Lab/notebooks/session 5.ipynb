```python
import gensim.downloader as api
from gensim.models import Word2Vec
```

```python
model = api.load("glove-wiki-gigaword-100")
```

```python
for word, result in model.most_similar("king"):
    print(word, ": ",round(result, 2))
```

```python
print(model.most_similar(positive=["king","woman"], negative=["man"], topn=1)) # [('father', 0.6721858382225037)]
```

```python
def convolve2d(image:np.array,
               kernel:np.array) ->np.array:
    s = 1
    # p = 0 
    output_size = ((image.shape[0] - kernel.shape[0]) // s ) + 1 
    output = np.zeros((output_size, output_size))
    for r in range(output_size):
        for c in range(output_size):
            window = image[r: r + 3, c: c + 3]
            output[r,c] = np.sum(window * kernel)
    return output

print(output)
```

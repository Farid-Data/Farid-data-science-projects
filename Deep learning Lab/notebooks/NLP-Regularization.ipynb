```python
import re
```

```python
text = "I'm learning NLP with python 3"
re.findall(r"\w+", text)
```

```python
text = "I have 3 cats, 12 dogs, and 2026 reasons to learn NLP."
print(re.findall(r"\d", text))
print(re.findall(r"\d+", text))
```

```python
text = "I    am   learning\nNLP\twith   Python."
print(re.sub(r"\s"," ", text)) # This return 'I    am   learning NLP with   Python.' but expected return is 'I am learning NLP with Python.'
print(re.sub(r"\s+", " ", text)) # This retuens the currecr result
```

```python
text = "Hello, NLP! I'm  , ,, , , , , learning Python. It's awesome!!!"
text = re.sub(r"[^\w\s]", "", text)
print(text) # returns = Hello NLP Im        learning Python Its awesome
re.sub(r"\s+", " ",text) # returns 'Hello NLP Im learning Python Its awesome'
```

```python
text = "Python 3.12 was released in 2023, and version 4 is coming."
re.findall(r"[^0-9]+", text) # I dicovered it my self, result: ['Python ', '.', ' was released in ', ', and version ', ' is coming.']
re.findall(r"[0-9]", text) # returns: ['3', '1', '2', '2', '0', '2', '3', '4']
```

```python
text = "Hello, NLP! I'm learning Python 3."
re.findall(r"[^\w]", text) # returns [',', ' ', '!', ' ', "'", ' ', ' ', ' ', '.']
re.findall(r"[^\w\s]", text) # here I decided to omit whitespcae so I did it :)
```

```python
text = "Hello, NLP! I'm learning Python 3."
re.findall(r"\W", text)    # returns: [',', ' ', '!', ' ', "'", ' ', ' ', ' ', '.']
re.findall(r"[^\w]", text) # returns: [',', ' ', '!', ' ', "'", ' ', ' ', ' ', '.']
print(re.findall(r"\W+", text)) # [', ', '! ', "'", ' ', ' ', ' ', '.'] less values cause it combines , + whitespace but previous patterns don't
print("""
Outputs are similar but pattern are diffrent :)
""")
```

```python
text = "I like Python"
print("+: ",re.findall(r"\w+", text)) # one and more 
print("*: ",re.findall(r"\w*", text)) # Zero and more (whitespace is not a word which is zero that's why we get it back)
```

```python
text = "color colour colr"

re.findall(r"colou?r", text)
```

```python
text = "cat cats dog dogs"
re.findall(r"cats?", text) # returns (cat, cats)
re.findall(r"dogs?", text) # return (dog, dogs)
```

```python
text = "My codes are 123, 4567, 89, and 2026."
re.findall(r"\w{4}", text) # ['code', '4567', '2026']
re.findall(r"\d{4}", text) # ['4567', '2026']
text1 = [int(i) for i in (re.findall(r"\d{4}", text))]  # [4567, 2026]
```

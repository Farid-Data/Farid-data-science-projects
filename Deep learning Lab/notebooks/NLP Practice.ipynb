```python
import nltk
from nltk import WordNetLemmatizer, sent_tokenize, word_tokenize
from nltk.corpus import stopwords
from sklearn.feature_extraction.text import CountVectorizer
```

```python
nltk.download("punkt")
nltk.download("punkt_tab")
nltk.download("stopwords")
nltk.download("wordnet")
```

```python
text = 'AI is changing the way we work. In 2026, NLP systems can analyze emails, websites, and social media posts! But can computers really understand human language? For example, "It\'s amazing," she said. Companies spend $1,299.99 on AI tools, and developers visit https://example.com to learn more. #MachineLearning is everywhere.'
```

```python
sentences = sent_tokenize(text)
for i in range(len(sentences)):
    print(f"Sentence number {i + 1}: ",sentences[i])
```

```python
tokens = word_tokenize(text)
for w in range(len(tokens)):
    print(f"word number {w + 1}: {tokens[w]}")
```

```python
stop_words = stopwords.words("english")
```

```python
content_words = [token for token in tokens
                 if token.isalpha() and token not in stop_words]
```

```python
stop_words_found = [token for token in tokens
                    if token.isalpha() and token in stop_words]
```

```python
stemmer = nltk.PorterStemmer()

stemmed_words = [stemmer.stem(word) for word in content_words]
```

```python
documents = [
    "I love machine learning and NLP.",
    "Machine learning is powerful.",
    "NLP helps computers understand human language.",
    "I love learning new things about AI."
]
```

```python
count_vectorizer = CountVectorizer()
```

```python
document_matrix = count_vectorizer.fit_transform(documents)
```

```python
print(count_vectorizer.get_feature_names_out())
```

```python
count_vectorizer.vocabulary_
```

```python
document_matrix.shape #Reurns (4, 16) that means there are four sentences and 16 words
```

```python
doc_array = document_matrix.toarray()
```

```python
import pandas as pd

matrix_df = pd.DataFrame(doc_array, columns =count_vectorizer.get_feature_names_out())
matrix_df
```

```python
test_text = ["I love machine learning and deep leanring a lot",
             "I wish I could start my personal business ASAP",
             "computers helps me learn nlp"]

print(count_vectorizer.transform(test_text))
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
tfidf_vectorizer = TfidfVectorizer()
tfidf_matrix = tfidf_vectorizer.fit_transform(documents)
```

```python
tfidf_vectorizer.vocabulary_
```

```python
feature_names  = tfidf_vectorizer.get_feature_names_out()
print(feature_names )
```

```python
tfidf_df = pd.DataFrame(tfidf_matrix.toarray(), columns = feature_names).round(2)
tfidf_df
```

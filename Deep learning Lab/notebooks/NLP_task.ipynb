```python
text = """Tech giant OpenAI released a new language model on 2026-08-15.
The model was trained on 500 billion tokens and scored 92.5 on benchmarks.
Read more at https://example.com/news/12345 or visit www.technews.com.
Experts say the model outperforms previous models in 87% of tasks.
The company plans to release 3 smaller models by 2027.
More details: https://example.com/details?id=678"""
```

1. **Remove URLs using Regex**

   * Remove links beginning with `https://...`
   * Remove links beginning with `www...`


```python
import re

print(re.sub(r"\bhttps://\S+\b","", text)) # removes https://
```

```python
print(re.sub(r"\bwww.\S+\b","",text)) # Removes Links contain www.
```

2. **Remove All Numbers**

   * Remove integers.
   * Remove decimals.
   * Remove percentages.
   * Remove dates.

```python
print(re.sub(r"\b\d+\b", "",text)) # Removes integers r"\b\d+\b"
```

```python
print(re.sub(r"\b\d+\.\d+\b", "", text))    # Removes Decimals ==> r"\b\d+\.\d+\b" , "", 
```

```python
print(re.sub(r"\b\%", "", text)) # Removes percentage
```

```python
print(re.sub(r"\d{4}-\d{2}-\d{2}", "", text)) # removes Dates ==> r"\b\d{4}-\d{2}\d{2}\b
```

```python
link_text = re.sub(r"\bhttps://\S+\b", "", text)
www_text = re.sub(r"\bwww\S+\b", "", link_text)
integer_text = re.sub(r"\b\d+\b", "", www_text)
decimal_text = re.sub(r"\b\d+\.\d+\b", "", integer_text)
percentage_text = re.sub(r"\b%", "", decimal_text)
final_text = re.sub(r"\b\d{4}-\d{2}-\d{2}\b", "", percentage_text)
print(final_text)
```

3. **Tokenize the Text**

   * Use `word_tokenize` to split the cleaned text into individual tokens.

```python
import nltk

tokens = nltk.word_tokenize(final_text)
tokens
```

4. **Clean the Tokens**

   * Convert all tokens to lowercase.
   * Remove punctuation.
   * Remove stopwords.

```python
lowercase_tokens = [token.lower() for token in tokens] # convert all tokens to lowercase
lowercase_tokens[:10]
```

```python
punctuation_removed_tokens = [re.sub(r"[^\w\d]", "", token) for token in lowercase_tokens]
punctuation_removed_tokens = [token for token in punctuation_removed_tokens if token]
punctuation_removed_tokens
```

```python
from nltk.corpus import stopwords

stop_words = stopwords.words("english")

cleaned_tokens = [token for token in punctuation_removed_tokens if token not in stop_words]

cleaned_tokens
```

5. **Find the Most Frequent Words**

   * Use `collections.Counter`.
   * Print the **10 most frequent words**.

```python
from collections import Counter

print(Counter(cleaned_tokens).most_common()[:10])
```

# Text Preprocessing Exercise

## Tasks

Given the text below, perform the following preprocessing operations:

1. **Convert to Lowercase**

   * Convert all letters in the text to lowercase English letters.

2. **Remove URLs**

   * Remove all URLs that start with `http`.
   * Remove all URLs that start with `www`.

3. **Remove Numbers**

   * Remove all numbers from the text.

4. **Remove Unwanted Characters**

   * Remove every character except:

     * Lowercase English letters (`a-z`)
     * Spaces

5. **Normalize Whitespace**

   * Replace consecutive spaces, tabs, and newlines with a single space.

6. **Strip Leading and Trailing Spaces**

   * Remove spaces from the beginning and end of the text.

7. **Display the Final Output**

   * Print or display the final cleaned text.


```python
text = """
Dr. Farshid Shirafkan is an experienced educator, programmer, and AI content creator with over 30 years of experience in teaching and software development. 
He is deeply passionate about teaching Machine Learning and Deep Learning, with a strong focus on making complex AI concepts practical, clear, 
and accessible for learners. 
Through his educational content and professional activities, Dr. Shirafkan has reached more than 600,000 learners on Faradars, one of Iran's leading online education platforms. Faradars website address: https://faradrs. org.
He has also achieved 25th place in LinkedIn AI content creation, reflecting his meaningful impact and consistent contribution to the AI education 
Their email address is fshirafkan51@gmail, and their phone number is 09121972028.
"""
```

1. **Convert to Lowercase**

   * Convert all letters in the text to lowercase English letters.

```python
lowercase_text = re.sub(r"[A-Z]", lambda match: match.group().lower(), text)
print(lowercase_text )
```

2. **Remove URLs**

   * Remove all URLs that start with `http`.
   * Remove all URLs that start with `www`.

```python
urls_removed_text = re.sub(r"\bhttps://\S+\b", "", lowercase_text)
all_urls_removed_text = re.sub(r"\bwww\.\S+\b", "", urls_removed_text)
print(all_urls_removed_text)
```

3. **Remove Numbers**

   * Remove all numbers from the text.

```python
numbers_removed_text = re.sub(r"\d+", "",  all_urls_removed_text)
print(numbers_removed_text)
```

4. **Remove Unwanted Characters**

   * Remove every character except:

     * Lowercase English letters (`a-z`)
     * Spaces

```python
cleaned_characters_text = re.sub(r"[^a-z\s]+", "", numbers_removed_text)
print(cleaned_characters_text)
```

5. **Normalize Whitespace**

   * Replace consecutive spaces, tabs, and newlines with a single space.

```python
normalized_text = re.sub(r"\s+", " ", cleaned_characters_text)
print(normalized_text)
```

6. **Strip Leading and Trailing Spaces**

   * Remove spaces from the beginning and end of the text.

```python
final_text = re.sub(r"^\s+|\s+$", "", normalized_text)
print(final_text)
```

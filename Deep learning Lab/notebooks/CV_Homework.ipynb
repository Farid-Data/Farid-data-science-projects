```python
import cv2
import matplotlib.pyplot as plt
import numpy as np
```

```python
image = np.full((500, 500, 3), 0, dtype=np.float32 )
mouth=cv2.line(image,(200, 300), (250, 300), (225,225, 225), thickness=3)

left_eye = cv2.circle(image, (175, 220), 10, (225,225, 225), 1)
right_eye = cv2.circle(image,(275, 220), 10, (225,225, 225), 1)
face = cv2.circle(image, (220,240), 100, (225,225, 225), 2)
nose = cv2.line(image, (225, 220), (225, 250),(225,225, 225),1 )
```

```python
plt.imshow(image, cmap="gray")
plt.axis("off")
plt.show()
```

```python
vertical_kernel = np.array([
    [-1, 2, -1],
    [-1, 2, -1],
    [-1, 2, -1],
])
```

```python
horizontal_kernel = np.array([
    [-1, -1, -1],
    [ 2,  2,  2],
    [-1, -1, -1],
])
```

```python
plt.imshow(cv2.filter2D(image, -1, vertical_kernel), cmap="gray")
plt.axis("off")
plt.show()
```

```python
plt.imshow(cv2.filter2D(image, -1, horizontal_kernel))
plt.axis("off")
plt.show()
```

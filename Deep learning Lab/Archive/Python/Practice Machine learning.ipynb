```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.datasets import load_diabetes
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
from sklearn.linear_model import LinearRegression, Lasso, Ridge , ElasticNet
from sklearn.pipeline import Pipeline
```

```python
data = load_diabetes()
X, y, feature_names = data.data, data.target, data.feature_names
```

```python
#  This God damn cell is super extra I did it because I wanted 😁
df = pd.DataFrame(X, columns=feature_names)
df["Target"] = y
```

```python
df.head(5)
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, random_state=42, test_size= 0.2)
X_train.shape, X_test.shape, y_train.shape, y_test.shape
```

```python
Line_reg_model = LinearRegression()
Line_reg_model.fit(X_train, y_train)
```

```python
lin_reg_pre = Line_reg_model.predict(X_test)
```

```python
mse_LR = mean_squared_error(y_test, lin_reg_pre)
mae = mean_absolute_error(y_test, lin_reg_pre)
r2 = r2_score(y_test, lin_reg_pre)
rmse_LR = np.sqrt(mse_LR)
print(f"MSE : {mse_LR:.2f}")
print(f"MAE : {mae:.2f}")
print(f"R2 score: {r2:.2f}")
print(f"RMSE : {rmse_LR:.2f}")
```

```python
param_grid = {"alpha": np.logspace(-6, 10, 100)}
ridge_model = GridSearchCV(Ridge(),param_grid, cv= 5,scoring="neg_mean_squared_error" )
ridge_model.fit(X_train, y_train)
```

```python
print(ridge_model.best_params_["alpha"])
print(ridge_model.best_score_)
print(ridge_model.best_estimator_)
```

```python
model = ridge_model.best_estimator_ # this is the best equation
ridge_pre = model.predict(X_test)
```

```python
mse_ridge = mean_squared_error(y_test, ridge_pre)
mae = mean_absolute_error(y_test,ridge_pre)
r2 = r2_score(y_test, ridge_pre)
rmse_ridge = np.sqrt(mse_ridge)
print(f"MSE : {mse_ridge:.5f}")
print(f"MAE : {mae:.5f}")
print(f"R2 score: {r2:.5f}")
print(f"RMSE : {rmse_ridge:.5f}")
```

```python
mse = mean_squared_error(y_test, lin_reg_pre)
mae = mean_absolute_error(y_test, lin_reg_pre)
r2 = r2_score(y_test, lin_reg_pre)
rmse = np.sqrt(mse)
print(f"MSE : {mse:.5f}")
print(f"MAE : {mae:.5f}")
print(f"R2 score: {r2:.5f}")
print(f"RMSE : {rmse:.5f}")
```

```python
# Lasso FUCK THE ENTIRE COUNTRY 🤬
param_grid_L1 = {"alpha": np.logspace(-4,10, 100)}
lasso_model = GridSearchCV(Lasso(max_iter=10000), param_grid_L1 , cv = 5 , scoring="neg_mean_squared_error")
lasso_model.fit(X_train, y_train)
lasso_best_model  = lasso_model.best_estimator_
lass_predict = lasso_best_model.predict(X_test)

mse_Lasso = mean_squared_error(y_test, lass_predict)
mae = mean_absolute_error(y_test, lass_predict)
r2 = r2_score(y_test, lass_predict)
rmse_Lasso = np.sqrt(mse)
print(f"MSE : {mse:.5f}")
print(f"MAE : {mae:.5f}")
print(f"R2 score: {r2:.5f}")
print(f"RMSE : {rmse:.5f}")
```

```python
# Elastic Regularization

Elastic_hyper = {"alpha": np.logspace(-4, 10 , 100),
                 "l1_ratio": [0.1,0.5,0.7,0.95, 0.99, 1]}
elastic_search = GridSearchCV(ElasticNet(max_iter=10000), Elastic_hyper, cv = 5 , scoring="neg_mean_squared_error")

elastic_search.fit(X_train, y_train)
elastic_best_model = elastic_search.best_estimator_
elastic_predict = elastic_best_model.predict(X_test)

mse_ELA = mean_squared_error(y_test, elastic_predict)
mae = mean_absolute_error(y_test, elastic_predict)
r2 = r2_score(y_test, elastic_predict)
rmse_ELA = np.sqrt(mse)
print(f"MSE : {mse:.5f}")
print(f"MAE : {mae:.5f}")
print(f"R2 score: {r2:.5f}")
print(f"RMSE : {rmse:.5f}")
```
